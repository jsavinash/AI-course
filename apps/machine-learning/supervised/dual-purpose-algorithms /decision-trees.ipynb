# Decision Trees

A decision tree splits the feature space with **axis-aligned** rules learned from the data. At each node it picks the split that most reduces impurity (e.g. Gini):

$$\text{Gini} = 1 - \sum_{k} p_k^{2}$$

Trees are **dual-purpose**: `DecisionTreeClassifier` for classification, `DecisionTreeRegressor` for regression.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the NB / LDA / QDA examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate


```python
tree = DecisionTreeClassifier(max_depth=3, random_state=0).fit(X, y)
preds = tree.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary and the learned tree


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = tree.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
axes[0].contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
axes[0].scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
axes[0].set_title("Decision boundary (axis-aligned splits)")
axes[0].set_xlabel("Feature 1"); axes[0].set_ylabel("Feature 2")

plot_tree(tree, filled=True, max_depth=3, fontsize=7, ax=axes[1])
axes[1].set_title("Learned tree")
plt.tight_layout()
plt.show()
```

## Takeaways

- Easy to interpret; the boundary is a set of rectangular regions.
- Prone to **overfitting** with deep trees -> control `max_depth` / `min_samples_leaf`.
- Ensemble methods (Random Forest, GBM) reduce that variance.
