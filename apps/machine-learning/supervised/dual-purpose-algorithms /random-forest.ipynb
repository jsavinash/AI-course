# Random Forest

A Random Forest is an **ensemble** of decision trees. Each tree is trained on a bootstrap sample of the data and a random subset of features; final predictions are averaged (regression) or majority-voted (classification):

$$\hat{y} = \text{mode/median}\big(T_1(x), T_2(x), \dots, T_B(x)\big)$$

Bagging many decorrelated trees greatly reduces the variance of a single tree.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the tree / KNN examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate


```python
rf = RandomForestClassifier(n_estimators=100, random_state=0).fit(X, y)
preds = rf.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
print("\nFeature importances:", np.round(rf.feature_importances_, 3))
```

## Decision boundary


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = rf.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 4.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("Random Forest decision boundary")
plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- Much more robust than a single tree (lower variance).
- Gives **feature importances** for free.
- Generally needs little tuning; `n_estimators` and `max_depth` are the main knobs.
