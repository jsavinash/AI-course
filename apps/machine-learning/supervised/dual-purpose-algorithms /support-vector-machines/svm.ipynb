# Support Vector Machine (SVM) - Classification

SVM finds the **maximum-margin** hyperplane that separates the classes. With the kernel trick it maps data into a higher dimension so non-linear boundaries become linear:

$$f(x) = \sum_i \alpha_i y_i K(x_i, x) + b$$

Common kernels: `linear`, `rbf` (Gaussian), `poly`. The `C` parameter trades off margin size vs. training errors.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the other dual-purpose examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate (RBF kernel)


```python
svm = SVC(kernel="rbf", C=1.0, gamma="scale").fit(X, y)
preds = svm.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = svm.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 4.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("SVM (RBF) decision boundary")
plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- SVM maximizes the margin between classes (good generalization).
- The **kernel trick** lets it learn non-linear boundaries cheaply.
- `SVR` is the regression counterpart (see `support-vector-machines/svr.ipynb`).
