# Support Vector Regression (SVR)

SVR is the regression version of SVM. Instead of a margin that separates classes, it fits a tube of width $\varepsilon$ and only penalizes points **outside** the tube:

$$\min_{w,b}\; \tfrac{1}{2}\|w\|^2 + C\sum_i \big(|y_i - f(x_i)| - \varepsilon\big)_+$$

With the RBF kernel it can model smooth non-linear curves. `C` controls tolerance for errors, $\varepsilon$ the tube width.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score

# 1-D non-linear regression: y = sin(x) + noise
np.random.seed(0)
x = np.linspace(0, 10, 40).reshape(-1, 1)
y = np.sin(x.ravel()) + np.random.randn(40) * 0.2
```

## Train and evaluate (RBF kernel)


```python
svr = SVR(kernel="rbf", C=10.0, gamma="scale", epsilon=0.1).fit(x, y)
y_pred = svr.predict(x)

print(f"MSE : {mean_squared_error(y, y_pred):.4f}")
print(f"R2  : {r2_score(y, y_pred):.4f}")
```

## Fit curve


```python
x_line = np.linspace(0, 10, 200).reshape(-1, 1)
y_line = svr.predict(x_line)

plt.figure(figsize=(7, 4.5))
plt.scatter(x.ravel(), y, color="crimson", label="Data", zorder=5)
plt.plot(x_line.ravel(), y_line, color="teal", linewidth=2.5, label="SVR fit")
plt.title("Support Vector Regression (RBF)")
plt.xlabel("x"); plt.ylabel("y")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- SVR fits a smooth curve within an $\varepsilon$-tube; outliers outside are penalized.
- `C` and `gamma` tune fit tightness and influence radius.
- The classification counterpart is `SVC` (see `svm.ipynb`).
