# LightGBM (Light Gradient Boosting Machine)

LightGBM is a fast gradient-boosting framework from Microsoft. Key idea: **leaf-wise** tree growth (splits the leaf with the largest loss reduction) instead of level-wise, plus histogram-based binning for speed:

$$\hat{y} = \sum_{m=1}^{M} \eta\, f_m(x), \quad f_m \text{ fit to gradients}$$

Leaf-wise growth is often faster and more accurate, but can overfit on small data (use `num_leaves` / `min_data_in_leaf` to control).


```python
import numpy as np
import matplotlib.pyplot as plt
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the other dual-purpose examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate

LightGBM is pre-installed in Colab; otherwise run `!pip install lightgbm`.


```python
lgbm = LGBMClassifier(n_estimators=100, learning_rate=0.1, random_state=0).fit(X, y)
preds = lgbm.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = lgbm.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 4.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("LightGBM decision boundary")
plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- Very fast and memory-efficient (histogram binning, leaf-wise growth).
- Great for large datasets; watch overfitting with `num_leaves`.
- Same gradient-boosting family as XGBoost, with different growth strategy.
