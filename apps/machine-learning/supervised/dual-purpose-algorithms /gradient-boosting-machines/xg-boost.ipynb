# XGBoost (Extreme Gradient Boosting)

XGBoost is an optimized, scalable implementation of gradient boosting. It adds **regularization** to the loss:

$$\mathcal{L} = \sum_i l(\hat{y}_i, y_i) + \sum_t \Omega(f_t), \qquad \Omega(f) = \gamma T + \tfrac{1}{2}\lambda \|w\|^2$$

This controls tree complexity ($T$ = leaves, $w$ = leaf weights), reducing overfitting and speeding training.


```python
import numpy as np
import matplotlib.pyplot as plt
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the other dual-purpose examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate

XGBoost is pre-installed in Colab; otherwise run `!pip install xgboost`.


```python
xgb = XGBClassifier(n_estimators=100, learning_rate=0.1, random_state=0, use_label_encoder=False, eval_metric="logloss").fit(X, y)
preds = xgb.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = xgb.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 4.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("XGBoost decision boundary")
plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- Regularized gradient boosting -> strong accuracy with less overfitting.
- Handles missing values and is very fast (parallel / GPU friendly).
- Main knobs: `n_estimators`, `learning_rate`, `max_depth`, `reg_lambda`.
