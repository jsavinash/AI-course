# Gradient Boosting Machines (GBM)

Gradient Boosting builds trees **sequentially**. Each new tree corrects the residual errors of the previous ensemble:

$$F_m(x) = F_{m-1}(x) + \eta\, h_m(x)$$

where $h_m$ is fit to the **negative gradient** of the loss, and $\eta$ is the learning rate. This is *boosting* (reduces bias) versus Random Forest's bagging (reduces variance).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the other dual-purpose examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate


```python
gbm = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=0).fit(X, y)
preds = gbm.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = gbm.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 4.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("Gradient Boosting decision boundary")
plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- Boosting adds trees one at a time to fix previous mistakes (low bias).
- `learning_rate` and `n_estimators` trade off speed vs. performance.
- Strong default model for tabular data; XGBoost / LightGBM / CatBoost are faster, optimized variants.
