# CatBoost (Categorical Boosting)

CatBoost is a gradient-boosting library from Yandex that handles **categorical features** natively (using ordered target statistics) and reduces **prediction shift** with an oblivious (symmetrical) tree structure:

$$\hat{y} = \sum_{m=1}^{M} \eta\, f_m(x), \quad f_m \text{ fit to gradients}$$

It often works well with minimal tuning and is robust on small/medium tabular data.


```python
import numpy as np
import matplotlib.pyplot as plt
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset (same as the other dual-purpose examples)
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate

CatBoost is NOT pre-installed in Colab; run `!pip install catboost` first.


```python
cb = CatBoostClassifier(n_estimators=100, learning_rate=0.1, random_state=0, verbose=0).fit(X, y)
preds = cb.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds.astype(int)))
```

## Decision boundary


```python
x1, x2 = X[:, 0], X[:, 1]
xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                     np.linspace(x2.min()-1, x2.max()+1, 200))
Z = cb.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 4.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("CatBoost decision boundary")
plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- Handles **categorical features** automatically (no manual encoding needed).
- Oblivious trees reduce prediction shift; often strong with default settings.
- Slower than LightGBM/XGBoost but very robust on small data.
