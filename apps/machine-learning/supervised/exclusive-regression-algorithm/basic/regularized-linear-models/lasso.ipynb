# Lasso Regression (L1 Regularization)

Lasso adds an **L1 penalty** (sum of absolute weights) to the least-squares cost:

$$J(\beta) = \|\mathbf{X}\beta - y\|^{2} + \alpha\,\|\beta\|_{1}$$

Because the L1 penalty is non-smooth at zero, Lasso can push **some coefficients exactly to 0** - giving automatic **feature selection**.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.preprocessing import StandardScaler

# Same synthetic dataset as the Ridge example
np.random.seed(42)
n = 30
size = np.random.uniform(500, 3000, n)
rooms = size / 400 + np.random.randn(n) * 1.5
noise_feat = np.random.randn(n)
X = np.column_stack([size, rooms, noise_feat])
y = 100 * size + 50 * rooms + np.random.randn(n) * 2000
feature_names = ["size", "rooms", "noise_feat"]
```

## Fit OLS vs. Lasso


```python
scaler = StandardScaler()
Xs = scaler.fit_transform(X)

ols = LinearRegression().fit(Xs, y)
lasso = Lasso(alpha=500.0, max_iter=10000).fit(Xs, y)

coef_df = pd.DataFrame({
    "feature": feature_names,
    "OLS": np.round(ols.coef_, 2),
    "Lasso (a=500)": np.round(lasso.coef_, 2),
})
print(coef_df.to_string(index=False))
print(f"\nIntercept  OLS   : {ols.intercept_:.2f}")
print(f"Intercept Lasso : {lasso.intercept_:.2f}")
print("\nFeatures dropped (coef = 0):",
      [f for f, c in zip(feature_names, lasso.coef_) if c == 0] or "none")
```

## Compare coefficient magnitudes


```python
x_pos = np.arange(len(feature_names))
width = 0.35

plt.bar(x_pos - width/2, ols.coef_, width, label="OLS", color="gray")
plt.bar(x_pos + width/2, lasso.coef_, width, label="Lasso (a=500)", color="crimson")
plt.xticks(x_pos, feature_names)
plt.ylabel("Coefficient (standardized)")
plt.title("Lasso can zero out coefficients (feature selection)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## Takeaways

- Lasso (**L1**) can set coefficients to **exactly 0**, so it performs feature selection.
- The irrelevant `noise_feat` is typically dropped, and the redundant `rooms` may be shrunk heavily.
- Larger `alpha` -> more coefficients forced to 0. Use `LassoCV` to choose `alpha`.
