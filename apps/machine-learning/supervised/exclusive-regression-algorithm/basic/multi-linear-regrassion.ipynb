# Multiple Linear Regression (Simple Example)

Predict a target from **several** features using the closed-form **Normal Equation**:

$$\hat{\boldsymbol{\beta}} = (\mathbf{X}^{\top}\mathbf{X})^{-1}\,\mathbf{X}^{\top}\mathbf{y}$$

so that $\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots$


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dataset: 5 houses. Features: [Size_SqFt, Bedrooms, Age_Years]
X = np.array([
    [1200, 2, 10],
    [1800, 3, 5],
    [2400, 4, 2],
    [1500, 3, 20],
    [3000, 4, 1],
], dtype=float)

y = np.array([210000, 290000, 380000, 230000, 450000], dtype=float)

feature_names = ["Size_SqFt", "Bedrooms", "Age_Years"]
df = pd.DataFrame(X, columns=feature_names)
df["Price_USD"] = y
print(df)
```

## Fit with the Normal Equation

We prepend a column of 1s to `X` for the intercept, then apply
$\boldsymbol{\beta} = (\mathbf{X}^{\top}\mathbf{X})^{-1}\mathbf{X}^{\top}\mathbf{y}$.


```python
X_design = np.hstack([np.ones((X.shape[0], 1)), X])  # add intercept column
beta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ y

intercept = beta[0]
coef = beta[1:]

print("=== LEARNED PARAMETERS ===")
print(f"Intercept (beta_0)            : {intercept:,.2f}")
for name, c in zip(feature_names, coef):
    print(f"{name:<12} coefficient        : {c:,.2f}")
```

## Performance Metrics & Residual Analysis


```python
y_pred = X_design @ beta
residuals = y - y_pred

mse = np.mean(residuals ** 2)
rmse = np.sqrt(mse)
ss_res = np.sum(residuals ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - ss_res / ss_tot

print("=== METRICS ===")
print(f"Mean Squared Error (MSE) : {mse:,.2f}")
print(f"Root MSE (RMSE)          : {rmse:,.2f}")
print(f"R-squared (R2)           : {r2:.4f}")

res_df = pd.DataFrame({
    "Actual": np.round(y, 1),
    "Predicted": np.round(y_pred, 1),
    "Residual": np.round(residuals, 1),
})
print()
print("=== PER-POINT RESIDUALS ===")
print(res_df.to_string(index=False))
```
