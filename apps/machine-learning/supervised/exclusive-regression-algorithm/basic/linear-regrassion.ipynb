# Basic Linear Regression

A minimal example of Ordinary Least Squares (OLS) simple linear regression.

We learn a straight line:

$$\hat{y} = m x + b$$

where `m` is the **slope** and `b` is the **intercept**.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dataset: Pizza Diameters (inches) and Prices ($)
X = np.array([6, 12], dtype=float)
Y = np.array([7.0, 14.0], dtype=float)
```

## Fit the line (OLS closed form)

$$m = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}, \qquad b = \bar{y} - m\bar{x}$$


```python
x_mean = np.mean(X)
y_mean = np.mean(Y)

numerator = np.sum((X - x_mean) * (Y - y_mean))
denominator = np.sum((X - x_mean) ** 2)

m = numerator / denominator
b = y_mean - m * x_mean

print(f"Slope (m)     : {m:.4f}")
print(f"Intercept (b) : {b:.4f}")
```

## Performance Metrics & Residual Analysis


```python
y_pred = m * X + b
residuals = Y - y_pred

mse = np.mean(residuals ** 2)
ss_res = np.sum(residuals ** 2)
ss_tot = np.sum((Y - y_mean) ** 2)
r2 = 1 - ss_res / ss_tot

print("=== METRICS ===")
print(f"Mean Squared Error : {mse:.4f}")
print(f"R-squared (R2)     : {r2:.4f}")

res_df = pd.DataFrame({
    "X (diameter)": X,
    "Y actual": Y,
    "Y predicted": np.round(y_pred, 3),
    "residual": np.round(residuals, 3),
})
print()
print("=== PER-POINT RESIDUALS ===")
print(res_df.to_string(index=False))
```

## Plot


```python
# Definitions copied from preceding cells to make this cell self-contained
X = np.array([6, 12], dtype=float)
Y = np.array([7.0, 14.0], dtype=float)

x_mean = np.mean(X)
y_mean = np.mean(Y)
numerator = np.sum((X - x_mean) * (Y - y_mean))
denominator = np.sum((X - x_mean) ** 2)
m = numerator / denominator
b = y_mean - m * x_mean

y_pred = m * X + b
residuals = Y - y_pred
# End of copied definitions

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Continuous x range so the regression line spans the axes
x_line = np.linspace(X.min() - 1, X.max() + 1, 100)
y_line = m * x_line + b

# Regression line
axes[0].scatter(X, Y, color="red", label="Actual", zorder=5)
axes[0].plot(x_line, y_line, color="blue", label="OLS line")
axes[0].set_title("Linear Regression")
axes[0].set_xlabel("Diameter (inches)")
axes[0].set_ylabel("Price ($)")
axes[0].legend()
axes[0].grid(True, linestyle="--", alpha=0.4)

# Residuals
axes[1].axhline(0, color="black", linewidth=1)
axes[1].scatter(X, residuals, color="green", s=80, zorder=5)
for xi, ri in zip(X, residuals):
  axes[1].annotate(f"{ri:.2f}", (xi, ri), textcoords="offset points", xytext=(0, 8), ha="center")
axes[1].set_title("Residuals")
axes[1].set_xlabel("Diameter (inches)")
axes[1].set_ylabel("Residual")
axes[1].grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
