# Naive Bayes (Gaussian)

Naive Bayes classifies using **Bayes' theorem** and the *naive* assumption that features are conditionally independent:

$$P(y \mid x) \propto P(y)\,\prod_{j=1}^{d} P(x_j \mid y)$$

With continuous features we use a **Gaussian** per feature per class:
$P(x_j \mid y) = \mathcal{N}(\mu_{j,y},\,\sigma_{j,y}^{2})$.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix

# 2-feature, 2-class dataset with different class shapes
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate


```python
model = GaussianNB().fit(X, y)
preds = model.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
def plot_boundary(model, X, y, title):
    x1, x2 = X[:, 0], X[:, 1]
    xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                         np.linspace(x2.min()-1, x2.max()+1, 200))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
    plt.title(title)
    plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
    plt.grid(True, linestyle="--", alpha=0.4)

plt.figure(figsize=(6, 4.5))
plot_boundary(model, X, y, "Naive Bayes decision boundary")
plt.tight_layout()
plt.show()
```

## Takeaways

- Fast, simple, works well with limited data and as a baseline.
- The "naive" independence assumption is often violated in practice but still works surprisingly well.
- Boundary here is roughly quadratic (Gaussian NB makes no linear assumption).
