# Quadratic Discriminant Analysis (QDA)

QDA is like LDA but lets **each class have its own covariance matrix**:

$$P(y=k \mid x) = \frac{\pi_k\,\mathcal{N}(x \mid \mu_k,\,\Sigma_k)}{\sum_j \pi_j\,\mathcal{N}(x \mid \mu_j,\,\Sigma_j)}$$

Because the covariances differ, the quadratic terms no longer cancel, giving a **curved (quadratic)** decision boundary. More flexible than LDA, but needs more data to estimate each $\Sigma_k$ reliably.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
from sklearn.metrics import accuracy_score, confusion_matrix

# Same 2-feature, 2-class dataset as the Naive Bayes / LDA examples
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate


```python
model = QuadraticDiscriminantAnalysis().fit(X, y)
preds = model.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
def plot_boundary(model, X, y, title):
    x1, x2 = X[:, 0], X[:, 1]
    xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                         np.linspace(x2.min()-1, x2.max()+1, 200))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
    plt.title(title)
    plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
    plt.grid(True, linestyle="--", alpha=0.4)

plt.figure(figsize=(6, 4.5))
plot_boundary(model, X, y, "QDA decision boundary (quadratic)")
plt.tight_layout()
plt.show()
```

## Takeaways

- QDA gives a **curved** boundary because each class has its own covariance.
- With our dataset (classes truly have different shapes) QDA often fits better than LDA.
- More parameters to estimate -> needs more data; otherwise can overfit.
