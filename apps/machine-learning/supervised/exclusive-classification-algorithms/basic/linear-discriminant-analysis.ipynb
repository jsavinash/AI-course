# Linear Discriminant Analysis (LDA)

LDA finds a **linear** decision boundary by assuming all classes share the **same covariance matrix** but have different means. It models:

$$P(y=k \mid x) = \frac{\pi_k\,\mathcal{N}(x \mid \mu_k,\,\Sigma)}{\sum_j \pi_j\,\mathcal{N}(x \mid \mu_j,\,\Sigma)}$$

Because the covariance $\Sigma$ is shared, the quadratic terms cancel and the boundary is a **straight line**.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score, confusion_matrix

# Same 2-feature, 2-class dataset as the Naive Bayes / QDA examples
np.random.seed(1)
n = 60
c0 = np.random.multivariate_normal([2, 3], [[2.0, 0], [0, 0.5]], n)
c1 = np.random.multivariate_normal([6, 4], [[0.5, 0], [0, 2.0]], n)
X = np.vstack([c0, c1])
y = np.array([0] * n + [1] * n)
```

## Train and evaluate


```python
model = LinearDiscriminantAnalysis().fit(X, y)
preds = model.predict(X)
print(f"Accuracy        : {accuracy_score(y, preds):.3f}")
print("Confusion matrix:")
print(confusion_matrix(y, preds))
```

## Decision boundary


```python
def plot_boundary(model, X, y, title):
    x1, x2 = X[:, 0], X[:, 1]
    xx, yy = np.meshgrid(np.linspace(x1.min()-1, x1.max()+1, 200),
                         np.linspace(x2.min()-1, x2.max()+1, 200))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.25, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k")
    plt.title(title)
    plt.xlabel("Feature 1"); plt.ylabel("Feature 2")
    plt.grid(True, linestyle="--", alpha=0.4)

plt.figure(figsize=(6, 4.5))
plot_boundary(model, X, y, "LDA decision boundary (linear)")
plt.tight_layout()
plt.show()
```

## Takeaways

- LDA produces a **straight-line** boundary (shared covariance assumption).
- Stable with few samples; also useful as a **dimensionality-reduction** step before other classifiers.
- If classes actually have very different shapes, a quadratic method (QDA) fits better.
